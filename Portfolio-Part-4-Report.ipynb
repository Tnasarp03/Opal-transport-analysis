## Portfolio-based Discussion Points

### 1. **Dataset Selection and Problem Identification**
For Portfolio 4, I chose the Opal dataset because it contains rich information on public transportation usage. This dataset provides insights into the number of trips, broken down by mode of transport and card type. It is highly relevant to current discussions on urban mobility, transportation optimization, and commuter behavior, making it a practical choice for analysis.

The problem I sought to address was understanding the factors that influence the number of trips taken by commuters. Specifically, I aimed to explore how variables such as `Card Type`, `Travel Mode`, and `Year_Month` affect trip volume. This problem is important because it can help transportation authorities with resource planning, managing fleet sizes, and optimizing services based on demand patterns. The overall objective was to build a predictive model that could estimate the number of trips based on these features.

### 2. **Model Selection and Suitability**
Three key models were used in the portfolio: **Linear Regression**, **Polynomial Regression**, and **K-Means Clustering**. The choice of these models was based on the problem's characteristics.

- **Linear Regression** was chosen as the baseline model due to its simplicity and interpretability. Since the dataset had a continuous numerical target (`Trip` counts), linear regression was a good starting point for understanding the linear relationship between features and the target.

- **Polynomial Regression** was used to capture any non-linearity present in the data. Transportation data often exhibit non-linear patterns (e.g., spikes in trip counts during rush hours), and polynomial regression helps model such relationships more effectively than linear regression.

- **K-Means Clustering** was chosen for identifying natural clusters in the dataset, specifically to understand how card types and transport modes group together in terms of trip behavior. Clustering allowed for the segmentation of commuters and their usage patterns, which can be useful for targeted interventions like promotional campaigns or service adjustments.

These models were suitable for the problem because they collectively addressed both predictive and descriptive aspects of the analysis. Linear and polynomial regression helped predict trip volumes, while K-Means clustering provided insights into user behavior by grouping similar travel patterns.

### 3. **Results and Conclusions**
One of the key insights from the study was that **bus transport mode** had the highest volume of trips, followed by trains. This finding aligned with my initial expectations, as buses often serve a more comprehensive network and tend to have greater accessibility compared to other modes like ferries or trams.

Another significant conclusion was that **adult card types** dominated trip counts. This was an expected outcome, considering that the majority of commuters are likely working adults. However, the dataset also revealed interesting trends for concession cards and student cards, suggesting a potential need for targeted services or discounts for these groups.

The regression models performed well, with polynomial regression providing a more accurate fit to the data compared to linear regression. However, the improvement was modest, indicating that while some non-linearity exists, the overall relationship between features and the target remains fairly linear.

The results from K-Means clustering were also insightful. By grouping trips based on card types and transport modes, the clusters revealed patterns in user preferences that could be leveraged for better resource allocation and marketing strategies.

Overall, the results were consistent with my intuitive expectations about public transport usage. The findings reinforced my understanding of urban mobility and provided actionable insights for optimizing transportation services.
